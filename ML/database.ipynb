```python
import os
import shutil
import random
from pathlib import Path
```

```python
source_dirs = {
    'coriandum_sativum': 'plant_img_database/coriandum_sativum',
    'mentha_spicata': 'plant_img_database/mentha_spicata',
    'echeveria_lilicana': 'plant_img_database/echeveria_lilicana',
    'fittonia_albivenis': 'plant_img_database/fittonia_albivenis',
    'salvia_officinales': 'plant_img_database/salvia_officinales',
    'haworthia_pumila': 'plant_img_database/haworthia_pumila',
}
```

```python
for plant_species, source_dir in source_dirs.items():
    for file_name in range(1, 101):
        src = os.path.join(source_dir, f"{file_name}.png")
        dst = os.path.join(source_dir, f"{plant_species}_{file_name}.png")
        os.rename(src, dst)
```

```python
random.seed(42)

for plant_species, source_dir in source_dirs.items():
    files = list(Path(source_dir).glob(f"{plant_species}_*.png"))
    random.shuffle(files)
    
    n_train = 70
    n_val = 15
    n_test = 15
    
    splits = {
        'plant_img_train': files[:n_train],
        'plant_img_val': files[n_train:n_train+n_val],
        'plant_img_test': files[n_train+n_val:],
    }
    
    for split, split_files in splits.items():
        target_dir = Path(split) / plant_species
        target_dir.mkdir(parents=True, exist_ok=True)
        
        for file_path in split_files:
            shutil.copy(file_path, target_dir / file_path.name)
```
