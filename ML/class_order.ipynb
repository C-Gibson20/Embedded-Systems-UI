```python
import torch
from torchvision.datasets import ImageFolder
import json
```

```python
ckpt_path = "saved_models/efficientnet_best2.ckpt"
ckpt = torch.load(ckpt_path, map_location="cpu")
print(ckpt.keys())
print(ckpt.get("hyper_parameters", {}))
```

```python
train_dir = "plant_img_train"
ds = ImageFolder(root=train_dir)

print("classes (idx -> name):")
for i, name in enumerate(ds.classes):
    print(i, name)

print("\nclass_to_idx:")
print(ds.class_to_idx)
```

```python
with open("class_to_idx.json", "w") as f:
    json.dump(ds.class_to_idx, f, indent=2)

print("Saved class_to_idx.json")
```
